```python
import matplotlib.pyplot as plt
import numpy as np

# Data
categories = [
    "Understood Purpose", 
    "Interested", 
    'Completed target task *',
    "Went to Froggy for Help", 
    "Found Froggy Useful", 
    "Used Froggy for Navigation"
]

general_public = [20, 11, 13, 13, 14, 2]
experts = [3, 1.5, 3, 2, 1.5,1]
max_scale = 30

# Positions and bar width
x = np.arange(len(categories))
width = 0.6

# Plotting
fig, ax = plt.subplots(figsize=(10, 6))
bars1 = ax.bar(x, general_public, width, label='General Public', color='#1f77b4')
bars2 = ax.bar(x, experts, width, bottom=general_public, label='Experts', color='orange')

# Add individual value labels
for i in range(len(x)):
    # Label for General Public
    ax.text(x[i], general_public[i] / 2, f'{general_public[i]}', ha='center', va='center', color='black')
    # Label for Experts
    ax.text(x[i], general_public[i] + experts[i] / 2, f'{experts[i]}', ha='center', va='center', color='black')

# Formatting
ax.set_ylabel("Responses")
ax.set_title("Key Walkthrough Observations (General Public and Experts)")
ax.set_xticks(x)
ax.set_xticklabels(categories, rotation=15)
ax.set_ylim(0, max_scale)
ax.legend()

plt.figtext(
    0.5, -0.02,
    "* Target task: General Public = reading the full story; Experts = going straight to data/projects",
    ha='center', fontsize=9, style='italic'
)
plt.tight_layout()
plt.show()
plt.savefig("Graphs/Walkthrough_findings.png", dpi=300)
```
